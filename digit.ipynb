```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
import os
```

```python

directory0 = "mnist-png/train/0"
directory1 = "mnist-png/train/1"
directory2 = "mnist-png/train/2"
directory3 = "mnist-png/train/3"
directory4 = "mnist-png/train/4"
directory5 = "mnist-png/train/5"
directory6 = "mnist-png/train/6"
directory7 = "mnist-png/train/7"
directory8 = "mnist-png/train/8"
directory9 = "mnist-png/train/9"


data = [[] for _ in range(10)]

for file in os.listdir(directory0):
    if file.endswith(".png"):
        img = Image.open(os.path.join(directory0, file)).convert("L")
        img = img.resize((28, 28))

        arr = np.array(img).flatten()
        data[0].append(arr)

data[0] = np.array(data[0])

print(f"0 :- {data[0].shape}")


for file in os.listdir(directory1):
    if file.endswith(".png"):
        img = Image.open(os.path.join(directory1, file)).convert("L")
        img = img.resize((28, 28))

        arr = np.array(img).flatten() 
        data[1].append(arr)

data[1] = np.array(data[1])

print(f"1 :- {data[1].shape}")



for file in os.listdir(directory2):
    if file.endswith(".png"):
        img = Image.open(os.path.join(directory2, file)).convert("L")
        img = img.resize((28, 28))

        arr = np.array(img).flatten() 
        data[2].append(arr)

data[2] = np.array(data[2])

print(f"2 :- {data[2].shape}")



for file in os.listdir(directory3):
    if file.endswith(".png"):
        img = Image.open(os.path.join(directory3, file)).convert("L")
        img = img.resize((28, 28))

        arr = np.array(img).flatten() 
        data[3].append(arr)

data[3] = np.array(data[3])

print(f"3 :- {data[3].shape}")



for file in os.listdir(directory4):
    if file.endswith(".png"):
        img = Image.open(os.path.join(directory4, file)).convert("L")
        img = img.resize((28, 28))

        arr = np.array(img).flatten() 
        data[4].append(arr)

data[4] = np.array(data[4])

print(f"4 :- {data[4].shape}")



for file in os.listdir(directory5):
    if file.endswith(".png"):
        img = Image.open(os.path.join(directory5, file)).convert("L")
        img = img.resize((28, 28))

        arr = np.array(img).flatten() 
        data[5].append(arr)

data[5] = np.array(data[5])

print(f"5 :- {data[5].shape}")



for file in os.listdir(directory6):
    if file.endswith(".png"):
        img = Image.open(os.path.join(directory6, file)).convert("L")
        img = img.resize((28, 28))

        arr = np.array(img).flatten() 
        data[6].append(arr)

data[6] = np.array(data[6])

print(f"6 :- {data[6].shape}")


for file in os.listdir(directory7):
    if file.endswith(".png"):
        img = Image.open(os.path.join(directory7, file)).convert("L")
        img = img.resize((28, 28))

        arr = np.array(img).flatten() 
        data[7].append(arr)

data[7] = np.array(data[7])

print(f"7 :- {data[7].shape}")



for file in os.listdir(directory8):
    if file.endswith(".png"):
        img = Image.open(os.path.join(directory8, file)).convert("L")
        img = img.resize((28, 28))

        arr = np.array(img).flatten() 
        data[8].append(arr)

data[8] = np.array(data[8])

print(f"8 :- {data[8].shape}")


for file in os.listdir(directory9):
    if file.endswith(".png"):
        img = Image.open(os.path.join(directory9, file)).convert("L")
        img = img.resize((28, 28))

        arr = np.array(img).flatten() 
        data[9].append(arr)

data[9] = np.array(data[9])

for k in range(10):
    data[k] = data[k].astype(np.float64) / 255.0

print(f"9 :- {data[9].shape}")
```

```python
#inialising weights for input and 2 hidden layer

for i in range(10):
    data[i] = np.array(data[i])

# Input layer to Hidden layer 1
w1 = np.random.randn(784, 128) * np.sqrt(2 / 784)
b1 = np.zeros((128))
layer1_Input = np.array([0]*784)

# Hidden layer 1 to Hidden layer 2
w2 = np.random.randn(128, 64) * np.sqrt(2 / 128)
b2 = np.zeros((64))
layer2_hidden = np.array([0]*128)

# Hidden layer 2 to Output layer
w3 = np.random.randn(64, 10) * np.sqrt(2 / 64)
b3 = np.zeros((10))
layer3_hidden = np.array([0]*64)

layer4_Output = [0]*10
```

```python
# Giving inputs in input layer

n = 0.001
epoch = 5
for _ in range(epoch):
    order = [(j, i) for j in range(10) for i in range(len(data[j]))]
    np.random.shuffle(order)
    for j, i in order:
        # printiny actual digit
        # print(f"Actual digit :- {j}")
        layer1_Input = data[j][i]
        y = np.array([0] * 10)
        y_hat = np.array([0]*10)
        y[j] = 1
        layer2_hidden = np.array([0] * 128)
        layer3_hidden = np.array([0] * 64)
        layer4_Output = np.array([0] * 10)

        #Input layer to first hidden layer
        layer2_hidden = np.dot(layer1_Input, w1) + b1
        ReluDerivative_of_z1 = layer2_hidden.copy()
        ReluDerivative_of_z1[ReluDerivative_of_z1 > 0] = 1
        ReluDerivative_of_z1[ReluDerivative_of_z1 <= 0] = 0
        ReluDerivative_of_z1 = ReluDerivative_of_z1.reshape(1,-1)


        #applying Relu activation fuction
        layer2_hidden = np.maximum(0, layer2_hidden)

        #first hidden layer to second hidden layer
        layer3_hidden = np.dot(layer2_hidden, w2) + b2
        ReluDerivative_of_z2 = layer3_hidden.copy()
        ReluDerivative_of_z2[ReluDerivative_of_z2 > 0] = 1
        ReluDerivative_of_z2[ReluDerivative_of_z2 <= 0] = 0
        ReluDerivative_of_z2 = ReluDerivative_of_z2.reshape(1,-1)

        #applying Relu activation fuction
        layer3_hidden = np.maximum(0, layer3_hidden)

        #second hidden layer to output layer
        layer4_Output = np.dot(layer3_hidden, w3) + b3

        # Applying softmax in output layer
        layer4_Output = layer4_Output.flatten()
        layer4_Output = layer4_Output - np.max(layer4_Output)
        layer4_Output = np.exp(layer4_Output)
        layer4_Output = layer4_Output / np.sum(layer4_Output)
            
            
        # Applying BackPropagation
        # Here we will use Sparse Categorical Cross-Entropy and Stochastic gradient descent
        y_hat = layer4_Output
        # print(y_hat)
        # printing predicted digit
        # print(f"pridicted digit = {np.argmax(y_hat)}")
        delta3 = y_hat - y
        delta3 = delta3.reshape(10, 1)
        a2 = layer3_hidden.copy()
        a2 = a2.reshape(1,-1)
        grad_w3_Loss = delta3 @ a2
        grad_w3_Loss = grad_w3_Loss.T
        prev_w3 = w3.copy()
        w3 = w3 - (n*grad_w3_Loss)
        b3 = b3 - (n*delta3.T)
            

        delta2 = prev_w3 @ delta3
        delta2  = delta2 * ReluDerivative_of_z2.T
        a1 = layer2_hidden.copy()
        a1 = a1.reshape(1,-1)
        grad_w2_Loss = delta2 @ a1
        grad_w2_Loss =  grad_w2_Loss.T
        prev_w2 = w2.copy()
        w2 = w2 - (n*grad_w2_Loss)
        b2 = b2 - (n*delta2.T)


        delta1 = prev_w2 @ delta2
        delta1  = delta1 * ReluDerivative_of_z1.T
        a0 = layer1_Input.copy()
        a0 = a0.reshape(1,-1)
        grad_w1_Loss = delta1 @ a0
        grad_w1_Loss =  grad_w1_Loss.T
        w1 = w1 - (n*grad_w1_Loss)
        b1 = b1 - (n*delta1.T)
```
